# 05 - SHAP Explainability

Computes SHAP values for the best Hybrid XGBoost, writes the Top-10 feature table (Table 4.8), and saves Figures 4.6 and 4.7.

```python
import sys, pathlib, joblib
sys.path.insert(0, str(pathlib.Path('.').resolve().parent))
from code import config, shap_explain

art = joblib.load(config.PROCESSED_DIR / 'best_hybrid_xgb.joblib')
pipe, names, X_test = art['pipe'], art['feature_names'], art['X_test']

# Pre-process via the pipeline up to the model (scaler only, since SMOTE is train-time)
X_pre = pipe[:-1].transform(X_test)
sv = shap_explain.explain_tree(pipe.named_steps['clf'], X_pre, names)

top = shap_explain.top_features_table(sv, top_k=12)
top.to_csv(config.PROCESSED_DIR / 'shap_top_hybrid.csv', index=False)
top
```

```python
shap_explain.save_summary_plot(sv, config.FIGURES_DIR / 'fig_4_6_shap_summary.png')
shap_explain.save_dependence_plot(sv, 'avg_sentiment', config.FIGURES_DIR / 'fig_4_7_shap_dependence.png')
```
