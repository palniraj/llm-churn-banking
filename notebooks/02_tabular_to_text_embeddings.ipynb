# 02 - Tabular -> Text Conversion + LLM Embeddings

Renders each bank-customer row into a natural-language description and pre-computes BERT and Sentence-Transformer embeddings (the latter is the default used in the headline numbers).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('.').resolve().parent))

from code import config, data_loader, preprocess, tabular_to_text, embeddings

bank = preprocess.clean_bank_churn(data_loader.load_bank_churn())
texts = tabular_to_text.render(bank, template='declarative')
print(texts.iloc[0])
```

```python
emb_st = embeddings.embed_sentence_transformer(texts)
embeddings.cache_embeddings(emb_st, 'bank_emb_st')
print('Sentence-Transformer embeddings:', emb_st.shape)
```
